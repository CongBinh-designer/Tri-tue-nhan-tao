```python
import copy
from collections import deque # Sử dụng deque để làm hàng đợi (Queue) hiệu quả

def printPuzzle(p):
    for row in p:
        print(row)
    print()

def findZero(puzzle):
    for i in range(3):
        for j in range(3):
            if puzzle[i][j] == 0:
                return i, j

def moveUp(puzzle):
    x, y = findZero(puzzle)

    if x == 0:
        return None

    newPuzzle = copy.deepcopy(puzzle)

    newPuzzle[x][y], newPuzzle[x-1][y] = \
    newPuzzle[x-1][y], newPuzzle[x][y]

    return newPuzzle

def moveDown(puzzle):
    x, y = findZero(puzzle)

    if x == 2:
        return None

    newPuzzle = copy.deepcopy(puzzle)

    newPuzzle[x][y], newPuzzle[x+1][y] = \
    newPuzzle[x+1][y], newPuzzle[x][y]

    return newPuzzle

def moveLeft(puzzle):
    x, y = findZero(puzzle)

    if y == 0:
        return None

    newPuzzle = copy.deepcopy(puzzle)

    newPuzzle[x][y], newPuzzle[x][y-1] = \
    newPuzzle[x][y-1], newPuzzle[x][y]

    return newPuzzle

def moveRight(puzzle):
    x, y = findZero(puzzle)

    if y == 2:
        return None

    newPuzzle = copy.deepcopy(puzzle)

    newPuzzle[x][y], newPuzzle[x][y+1] = \
    newPuzzle[x][y+1], newPuzzle[x][y]

    return newPuzzle

def solve_bfs(start_node, goal_node):
    queue = deque([(start_node, [])])

    visited = set()
    visited.add(tuple(tuple(row) for row in start_node))

    while queue:
        current_state, path = queue.popleft()

        if current_state == goal_node:
            return path + [current_state]

        possible_moves = [
            moveUp(current_state),
            moveDown(current_state),
            moveLeft(current_state),
            moveRight(current_state)
        ]

        for next_state in possible_moves:
            if next_state is not None:
                state_tuple = tuple(tuple(row) for row in next_state)

                if state_tuple not in visited:
                    visited.add(state_tuple)
                    queue.append((next_state, path + [current_state]))

    return None

goal = [[1, 2, 3], [4, 5, 6], [7, 8, 0]]
print("Nhap ma tran bat dau (3x3):")
start_puzzle = []
for i in range(3):
    start_puzzle.append(list(map(int, input().split())))

result_path = solve_bfs(start_puzzle, goal)

if result_path:
    print(f"\nTim thay loi giai sau {len(result_path) - 1} buoc:")
    for step_num, state in enumerate(result_path):
        print(f"Buoc {step_num}:")
        printPuzzle(state)
else:
    print("Khong tim thay loi giai.")
```
