```python
import random
import copy

room_original = []
m, n = map(int, input("Nhap so dong va cot: ").split())

for i in range(m):
    row = list(map(int, input().split()))
    room_original.append(row)

def printRoom(room, x, y):
    for i in range(m):
        for j in range(n):
            if x == i and y == j:
                print("A", end=" ")
            else:
                print(room[i][j], end=" ")
        print()

def is_clean(room):
    """Kiểm tra xem phòng đã sạch hoàn toàn chưa"""
    for i in range(m):
        for j in range(n):
            if room[i][j] == 1:
                return False
    return True

def get_neighbors(x, y):
    """Sinh ra các trạng thái lân cận hợp lệ"""
    neighbors = []
    if x > 0:
        neighbors.append((x - 1, y, "UP"))
    if y > 0:
        neighbors.append((x, y - 1, "LEFT"))
    if x < m - 1:
        neighbors.append((x + 1, y, "DOWN"))
    if y < n - 1:
        neighbors.append((x, y + 1, "RIGHT"))
    return neighbors

MAX_RESTART = 5
success = False
final_x, final_y = -1, -1
room = copy.deepcopy(room_original)

print("\n--- BẮT ĐẦU QUÁ TRÌNH LEO NÚI KHỞI ĐỘNG LẠI NGẪU NHIÊN ---")

for i in range(1, MAX_RESTART + 1):
    print(f"\n================ LƯỢT CHẠY RESTART THỨ {i} ================")

    x = random.randint(0, m - 1)
    y = random.randint(0, n - 1)

    step = 0
    max_steps_per_run = 20

    while step < max_steps_per_run:
        print(f"-> Step: {step} | Vi tri: ({x}, {y})")

        if room[x][y] == 1:
            print("[Hút bụi bẩn tại ô này]")
            room[x][y] = 0

        printRoom(room, x, y)
        print()

        if is_clean(room):
            print(f"-> THÀNH CÔNG TẠI LƯỢT {i}: Phòng đã sạch hoàn toàn!")
            success = True
            final_x, final_y = x, y
            break

        current_value = room[x][y]

        neighbors = get_neighbors(x, y)

        better_neighbors = []
        for next_x, next_y, direction in neighbors:
            next_value = room[next_x][next_y]
            if next_value > current_value:
                better_neighbors.append((next_x, next_y, direction))

        if len(better_neighbors) == 0:
            print(f"[Kẹt cực đại cục bộ tại ({x}, {y})! Thoát vòng lặp trong để chuyển sang lượt tiếp theo]")
            break

        else:
            next_x, next_y, direction = random.choice(better_neighbors)
            print(f"   Di chuyen sang huong tot nhat: {direction}")

            x, y = next_x, next_y

        step += 1

    if success:
        break

print("\n--- KẾT QUẢ CUỐI CÙNG ---")
if success:
    print(f"Thành công dọn sạch phòng ở vị trí kết thúc: ({final_x}, {final_y})")
else:
    print("TRẢ VỀ: 'Thất bại' (Đã thử hết số lượt MAX_RESTART nhưng không dọn sạch được các vết bẩn cô lập ở xa).")
```
