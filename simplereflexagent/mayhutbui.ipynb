```python
import random

room = []
m,n = map(int,input("Nhap so dong va cot: ").split())

x = random.randint(0, m-1)
y = random.randint(0, n-1)

for i in range(m):
    row = list(map(int, input().split()))
    room.append(row)

def printRoom(room,x,y):
    for i in range(m):
        for j in range(n):
            if x == i and y == j:
                print("A", end=" ")
            else:
                print(room[i][j], end=" ")
        print()

def possible_moves(x,y):
    possible_moves = []
    if x > 0:
        possible_moves.append("UP")
    if y > 0:
        possible_moves.append("LEFT")
    if x < m-1:
        possible_moves.append("DOWN")
    if y < n-1:
        possible_moves.append("RIGHT")

    return possible_moves
def is_clean():
    for i in range(m):
        for j in range(n):
            if room[i][j] == 1:
                return False
    return True

max_steps = 50
step = 0
while is_clean() == False and step < max_steps:
    print("Step:", step)
    if room[x][y] == 1:
        room[x][y] = 0
    printRoom(room,x,y)
    print()

    moves = possible_moves(x,y)
    move = random.choice(moves)

    if move == "UP":
        x-=1
    elif move == "LEFT":
        y-=1
    elif move == "DOWN":
        x+=1
    elif move == "RIGHT":
        y+=1

    step +=1
if is_clean() == False:
    print("Phong chua sach")
else:
    print("Phong da sach")
```

## PEAS:

S: Hàm để agent đọc dữ liệu
