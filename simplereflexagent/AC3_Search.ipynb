```python
def rm_inconsistent_values(Xi, Xj, domains, constraints_func):
    removed = False
    current_domain_Xi = list(domains[Xi])

    for x in current_domain_Xi:
        allows = False
        for y in domains[Xj]:
            if constraints_func(Xi, Xj, x, y):
                allows = True
                break

        if not allows:
            domains[Xi].remove(x)
            removed = True

    return removed


def ac3(variables, domains, neighbors, constraints_func):
    queue = []
    for Xi in variables:
        for Xj in neighbors[Xi]:
            queue.append((Xi, Xj))

    while queue:
        Xi, Xj = queue.pop(0)

        if rm_inconsistent_values(Xi, Xj, domains, constraints_func):

            if len(domains[Xi]) == 0:
                return False

            for Xk in neighbors[Xi]:
                if Xk != Xj and (Xk, Xi) not in queue:
                    queue.append((Xk, Xi))

    return True
variables = ['X', 'Y', 'Z']

domains = {
    'X': [1, 2, 3],
    'Y': [1, 2, 3],
    'Z': [2, 3]
}

neighbors = {
    'X': ['Y'],
    'Y': ['X', 'Z'],
    'Z': ['Y']
}

def my_constraints(Xi, Xj, val_i, val_j):
    if Xi == 'X' and Xj == 'Y': return val_i < val_j
    if Xi == 'Y' and Xj == 'X': return val_j < val_i
    if Xi == 'Y' and Xj == 'Z': return val_i < val_j
    if Xi == 'Z' and Xj == 'Y': return val_j < val_i
    return True

print("--- Trước khi chạy AC-3 ---")
print(domains)

has_solution = ac3(variables, domains, neighbors, my_constraints)

print("\n--- Kết quả sau khi chạy AC-3 ---")
print(f"Bài toán có nghiệm? {has_solution}")
print(f"Miền giá trị sau khi được thu gọn tối ưu: {domains}")
```
