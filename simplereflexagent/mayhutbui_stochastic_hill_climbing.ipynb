```python
import random

room = []
m, n = map(int, input("Nhap so dong va cot: ").split())

x = random.randint(0, m - 1)
y = random.randint(0, n - 1)

for i in range(m):
    row = list(map(int, input().split()))
    room.append(row)


def printRoom(room, x, y):
    for i in range(m):
        for j in range(n):
            if x == i and y == j:
                print("A", end=" ")
            else:
                print(room[i][j], end=" ")
        print()


def is_clean():
    """Hàm kiểm tra xem toàn bộ phòng đã sạch chưa"""
    for i in range(m):
        for j in range(n):
            if room[i][j] == 1:
                return False
    return True


def get_neighbors(x, y):
    """Sinh ra TẤT CẢ các trạng thái lân cận hợp lệ"""
    neighbors = []
    if x > 0:
        neighbors.append((x - 1, y, "UP"))
    if y > 0:
        neighbors.append((x, y - 1, "LEFT"))
    if x < m - 1:
        neighbors.append((x + 1, y, "DOWN"))
    if y < n - 1:
        neighbors.append((x, y + 1, "RIGHT"))
    return neighbors


step = 0
max_steps = 50

print("\n--- BẮT ĐẦU QUÁ TRÌNH LEO ĐỒI NGẪU NHIÊN ---")

while step < max_steps:
    print(f"Step: {step} (Vi tri hien tai: {x}, {y})")

    if room[x][y] == 1:
        print("-> Phát hiện bụi bẩn! Đang hút bụi...")
        room[x][y] = 0

    printRoom(room, x, y)
    print()

    if is_clean():
        print("-> Đạt mục tiêu: Phòng đã sạch hoàn toàn!")
        break

    current_value = room[x][y]

    neighbors = get_neighbors(x, y)

    better_neighbors = []
    for next_x, next_y, direction in neighbors:
        next_value = room[next_x][next_y]

        if next_value > current_value:
            better_neighbors.append((next_x, next_y, direction))

    if len(better_neighbors) == 0:
        print("-> NẾU Better_Neighbors RỖNG: Dừng vì đã đạt cực đại cục bộ!")
        break

    else:
        next_x, next_y, direction = random.choice(better_neighbors)
        print(f"-> Chon ngau nhien tu tap con tot hon: Huong {direction}")

        x, y = next_x, next_y

    step += 1

print("--- KẾT QUẢ ---")
if is_clean():
    print("Phong da sach hoàn toàn!")
else:
    print("Phong chua sach xong (Robot dung lai vi kẹt ở cực đại cục bộ).")
```
