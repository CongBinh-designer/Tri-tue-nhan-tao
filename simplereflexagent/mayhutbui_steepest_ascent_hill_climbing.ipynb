```python
import random

room = []
m, n = map(int, input("Nhap so dong va cot: ").split())

x = random.randint(0, m - 1)
y = random.randint(0, n - 1)

for i in range(m):
    row = list(map(int, input().split()))
    room.append(row)


def printRoom(room, x, y):
    for i in range(m):
        for j in range(n):
            if x == i and y == j:
                print("A", end=" ")
            else:
                print(room[i][j], end=" ")
        print()


def is_clean():
    """Hàm kiểm tra xem toàn bộ phòng đã sạch chưa"""
    for i in range(m):
        for j in range(n):
            if room[i][j] == 1:
                return False
    return True


def get_neighbors(x, y):
    """Sinh ra TẤT CẢ các trạng thái lân cận hợp lệ"""
    neighbors = []
    if x > 0:
        neighbors.append((x - 1, y, "UP"))
    if y > 0:
        neighbors.append((x, y - 1, "LEFT"))
    if x < m - 1:
        neighbors.append((x + 1, y, "DOWN"))
    if y < n - 1:
        neighbors.append((x, y + 1, "RIGHT"))
    return neighbors


step = 0
max_steps = 50

print("\n--- BẮT ĐẦU QUÁ TRÌNH LEO NÚI DỐC NHẤT ---")

while step < max_steps:
    print(f"Step: {step} (Vi tri hien tai: {x}, {y})")

    if room[x][y] == 1:
        print("-> Phát hiện bụi bẩn! Đang hút bụi...")
        room[x][y] = 0

    printRoom(room, x, y)
    print()

    if is_clean():
        print("-> Đạt mục tiêu: Phòng đã sạch hoàn toàn!")
        break

    current_value = room[x][y]

    neighbors = get_neighbors(x, y)

    best_neighbors_list = []
    max_neighbor_value = -float('inf')

    for next_x, next_y, direction in neighbors:
        next_value = room[next_x][next_y]

        if next_value > max_neighbor_value:
            max_neighbor_value = next_value
            best_neighbors_list = [(next_x, next_y, direction)]
        elif next_value == max_neighbor_value:
            best_neighbors_list.append((next_x, next_y, direction))

    best_neighbor = random.choice(best_neighbors_list) if best_neighbors_list else None

    if best_neighbor and max_neighbor_value > current_value:
        next_x, next_y, direction = best_neighbor
        print(f"-> Chon trạng thái lân cận tốt nhất: {direction} (Value: {max_neighbor_value})")
        x, y = next_x, next_y
    else:
        print("-> NGƯỢC LẠI: Trả về Current_State (Dừng vì đã đạt cực đại cục bộ)!")
        break

    step += 1

print("--- KẾT QUẢ ---")
if is_clean():
    print("Phong da sach!")
else:
    print("Phong chua sach xong (Robot dung lai vi bi ket o cuc dai cuc bo).")
```
