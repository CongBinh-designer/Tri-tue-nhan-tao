```python
import random

room = []
m,n = map(int,input("Nhap so dong va cot: ").split())

x = random.randint(0, m-1)
y = random.randint(0, n-1)
visited_positions = set()

for i in range(m):
    row = list(map(int, input().split()))
    room.append(row)

def save_position(x,y):
    visited_positions.add((x,y))

def is_visited(x,y):
    return (x,y) in visited_positions

def print_room(room,x,y):
    for i in range(m):
        for j in range(n):
            if x == i and y == j:
                print("A", end=" ")
            else:
                print(room[i][j], end=" ")
        print()

def is_clean():
    for i in range(m):
        for j in range(n):
            if room[i][j] == 1:
                return False
    return True

max_steps = 50
step = 0

while is_clean() == False and step < max_steps:
    print("Step:", step)
    save_position(x,y)

    if room[x][y] == 1:
        room[x][y] = 0

    print_room(room,x,y)
    print()

    moves = []
    if x > 0: moves.append((x - 1, y))
    if y > 0: moves.append((x, y - 1))
    if x < m - 1: moves.append((x + 1, y))
    if y < n - 1: moves.append((x, y + 1))

    unvisited_moves = [m for m in moves if not is_visited(m[0], m[1])]
    if unvisited_moves:
        next_move = random.choice(unvisited_moves)
    else:
        next_move = random.choice(moves)

    x = next_move[0]
    y = next_move[1]

    step +=1

if is_clean():
    print("Phong da sach")
else:
    print("Phong chua sach")
```
