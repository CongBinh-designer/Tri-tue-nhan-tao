```python
import random
import math

room = []
m, n = map(int, input("Nhap so dong va cot: ").split())

x = random.randint(0, m - 1)
y = random.randint(0, n - 1)

for i in range(m):
    row = list(map(int, input().split()))
    room.append(row)


def printRoom(room, x, y):
    for i in range(m):
        for j in range(n):
            if x == i and y == j:
                print("A", end=" ")
            else:
                print(room[i][j], end=" ")
        print()


def is_clean(room_state):
    """Kiểm tra xem phòng sạch hoàn toàn chưa"""
    for row in room_state:
        if 1 in row:
            return False
    return True


def h(room_state):
    """
    Tổng số ô bẩn còn lại trong phòng càng ít càng tốt.
    """
    return sum(row.count(1) for row in room_state)


def get_random_neighbor(cx, cy, current_room):
    """
    Trả về tọa độ mới và ma trận phòng giả lập sau khi robot đặt chân đến ô đó.
    """
    moves = []
    if cx > 0: moves.append((cx - 1, cy))      # UP
    if cx < m - 1: moves.append((cx + 1, cy))  # DOWN
    if cy > 0: moves.append((cx, cy - 1))      # LEFT
    if cy < n - 1: moves.append((cx, cy + 1))  # RIGHT

    next_x, next_y = random.choice(moves)

    next_room = [list(row) for row in current_room]
    if next_room[next_x][next_y] == 1:
        next_room[next_x][next_y] = 0

    return next_x, next_y, next_room


# Khởi tạo các tham số nhiệt độ
T = 100.0       # T0: Nhiệt độ ban đầu
Tmin = 0.1      # Tmin: Nhiệt độ tối thiểu để dừng thuật toán
alpha = 0.95    # Hệ số giảm nhiệt độ sau mỗi bước chạy

# current state = start
current_x, current_y = x, y
current_room = [list(row) for row in room]

if current_room[current_x][current_y] == 1:
    current_room[current_x][current_y] = 0

step = 0
print("\n--- BẮT ĐẦU THUẬT TOÁN SIMULATED ANNEALING ---")

# while T > Tmin:
while T > Tmin:
    print(f"\n================ STEP {step} (Nhiệt độ T = {T:.2f}) ================")
    print(f"Vị trí hiện tại: ({current_x}, {current_y}) | Số ô bẩn còn lại: {h(current_room)}")
    printRoom(current_room, current_x, current_y)
    print()

    # if current state == goal: return current state
    if is_clean(current_room):
        print("-> ĐẠT ĐÍCH (GOAL): Phòng đã được dọn dẹp sạch sẽ hoàn toàn!")
        break

    # next state = RandomNeighbor(current state)
    next_x, next_y, next_room = get_random_neighbor(current_x, current_y, current_room)

    # delta = h(next state) - h(current state)
    delta = h(next_room) - h(current_room)

    print(f"-> Chọn ngẫu nhiên ô tiếp theo: ({next_x}, {next_y})")
    print(f"-> Chi phí h(next) = {h(next_room)} | h(current) = {h(current_room)} | Δ = {delta}")

    # if delta < 0:
    if delta < 0:
        print("   [Δ < 0: Hướng đi tốt giúp giảm vết bẩn! Chấp nhận ngay]")
        # current state = next state
        current_x, current_y = next_x, next_y
        current_room = next_room
    # else:
    else:
        # p = exp(-delta / T)
        p = math.exp(-delta / T)
        random_val = random.uniform(0, 1)

        print(f"   [Δ >= 0: Hướng đi không làm giảm vết bẩn. Xác suất chấp nhận p = {p:.4f} | Số ngẫu nhiên = {random_val:.4f}]")

        # if Random(0,1) < p:
        if random_val < p:
            print("   => Số ngẫu nhiên < p: Chấp nhận đi lùi / đi dạo sang ô bẩn ít tối ưu để dò đường!")
            # current state = next state
            current_x, current_y = next_x, next_y
            current_room = next_room
        else:
            print("   => Từ chối bước đi này! Robot giữ nguyên vị trí cũ.")

    # T = alpha * T
    T = alpha * T
    step += 1

# return current state
print("\n--- KẾT QUẢ CUỐI CÙNG ---")
print(f"Thuật toán kết thúc tại vị trí robot: ({current_x}, {current_y})")
if is_clean(current_room):
    print("Trạng thái phòng trả về: SẠCH BONG!")
else:
    print(f"Trạng thái phòng trả về: CHƯA SẠCH (Còn lại {h(current_room)} ô bẩn do thuật toán bị hạ nhiệt hết mức).")
```
