```python
import random

room = []
m, n = map(int, input("Nhap so dong va cot: ").split())

x = random.randint(0, m - 1)
y = random.randint(0, n - 1)

for i in range(m):
    row = list(map(int, input().split()))
    room.append(row)


def printRoom(room, x, y):
    for i in range(m):
        for j in range(n):
            if x == i and y == j:
                print("A", end=" ")
            else:
                print(room[i][j], end=" ")
        print()


def get_neighbors(x, y):
    neighbors = []
    if x > 0:
        neighbors.append((x - 1, y, "UP"))
    if y > 0:
        neighbors.append((x, y - 1, "LEFT"))
    if x < m - 1:
        neighbors.append((x + 1, y, "DOWN"))
    if y < n - 1:
        neighbors.append((x, y + 1, "RIGHT"))
    return neighbors


def is_clean():
    for i in range(m):
        for j in range(n):
            if room[i][j] == 1:
                return False
    return True

max_steps = 50
step = 0

print("\n--- BẮT ĐẦU QUÁ TRÌNH LEO NÚI ---")

while step < max_steps:
    print(f"Step: {step} (Vi tri hien tai: {x}, {y})")

    if room[x][y] == 1:
        print("-> Phát hiện bụi bẩn! Đang hút bụi...")
        room[x][y] = 0

    printRoom(room, x, y)
    print()

    if is_clean():
        break

    current_value = room[x][y]

    neighbors = get_neighbors(x, y)
    found_better_move = False

    for next_x, next_y, move_direction in neighbors:
        next_value = room[next_x][next_y]

        if next_value > current_value:
            print(f"-> Chon huong tot hon: {move_direction}")
            x, y = next_x, next_y
            found_better_move = True
            break

    if not found_better_move:
        print("-> Dung vi dat cuc dai cuc bo (Xung quanh khong con o ban nao)!")
        break

    step += 1

print("--- KẾT QUẢ ---")
if is_clean():
    print("Phong da sach hoan toan!")
else:
    print("Phong chua sach (Robot bi ket o cuc dai cuc bo).")
```
