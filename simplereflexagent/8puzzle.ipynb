```python
import random
import copy

puzzle = []

for i in range(3):
    row = list(map(int, input().split()))
    puzzle.append(row)

def printPuzzle(p):
    for row in p:
        print(row)
    print()

def findZero(puzzle):
    for i in range(3):
        for j in range(3):
            if puzzle[i][j] == 0:
                return i, j

def moveUp(puzzle):
    x, y = findZero(puzzle)

    if x == 0:
        return None

    newPuzzle = copy.deepcopy(puzzle)

    newPuzzle[x][y], newPuzzle[x-1][y] = \
    newPuzzle[x-1][y], newPuzzle[x][y]

    return newPuzzle

def moveDown(puzzle):
    x, y = findZero(puzzle)

    if x == 2:
        return None

    newPuzzle = copy.deepcopy(puzzle)

    newPuzzle[x][y], newPuzzle[x+1][y] = \
    newPuzzle[x+1][y], newPuzzle[x][y]

    return newPuzzle

def moveLeft(puzzle):
    x, y = findZero(puzzle)

    if y == 0:
        return None

    newPuzzle = copy.deepcopy(puzzle)

    newPuzzle[x][y], newPuzzle[x][y-1] = \
    newPuzzle[x][y-1], newPuzzle[x][y]

    return newPuzzle

def moveRight(puzzle):
    x, y = findZero(puzzle)

    if y == 2:
        return None

    newPuzzle = copy.deepcopy(puzzle)

    newPuzzle[x][y], newPuzzle[x][y+1] = \
    newPuzzle[x][y+1], newPuzzle[x][y]

    return newPuzzle

visited = set()

current = puzzle

while True:
    state = tuple(tuple(row) for row in current)

    if state in visited:
        printPuzzle(current)
        print("Trang thai bi lap lai -> Dung")
        break

    visited.add(state)
    printPuzzle(current)
    possibleMoves = []
    up = moveUp(current)

    if up is not None:
        possibleMoves.append(up)

    down = moveDown(current)
    if down is not None:
        possibleMoves.append(down)

    left = moveLeft(current)
    if left is not None:
        possibleMoves.append(left)

    right = moveRight(current)
    if right is not None:
        possibleMoves.append(right)

    current = random.choice(possibleMoves)
```
